```python
import pandas as pd

from eda_utils import (
    basic_overview,
    plot_numerical_distributions,
    plot_categorical_counts,
    detect_outliers,
    correlation_heatmap
)

df = pd.read_csv("datasets/netflix_titles.csv")
df.head()

OUTPUT_DIR = "outputs_netflix"

basic_overview(df)

plot_categorical_counts(df, "type", OUTPUT_DIR)

plot_categorical_counts(df, "rating", OUTPUT_DIR)

df["duration_num"] = (
    df["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

plot_numerical_distributions(
    df[["release_year", "duration_num"]],
    OUTPUT_DIR
)

detect_outliers(
    df[["release_year", "duration_num"]],
    OUTPUT_DIR
)

correlation_heatmap(
    df[["release_year", "duration_num"]],
    OUTPUT_DIR
)

```
